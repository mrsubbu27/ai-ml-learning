## A4. Broadcasting

**Question:**

**Answer:**  
**Broadcasting** is a NumPy feature that allows arrays with different shapes to be used together in **element-wise operations**, when their shapes are compatible.

**General Rule:**

- **Same dimension:** Two dimensions are compatible if they are equal.  
  Example: `3` and `3` → compatible.

- **One dimension is 1:** Two dimensions are also compatible if one of them is `1`.  
  Example: `3` and `1` → compatible.

NumPy compares the dimensions from **right to left**. If every pair of dimensions satisfies one of these conditions, broadcasting is possible. Otherwise, NumPy cannot perform the element-wise operation.

**Example:**
```python

              [4, 5, 6]])

## A5. `flatten()` vs `ravel()`

**Question:**  
What is the difference between `flatten()` and `ravel()`? If you are working with large images and memory is constrained, which would you prefer and why? **[2 marks]**

**Answer:**  
Both `flatten()` and `ravel()` convert a multi-dimensional NumPy array into a **one-dimensional array**, but they handle memory differently.

- **`flatten()`** → Always creates a **copy** of the original array, so it uses additional memory.

- **`ravel()`** → Usually returns a **view** of the original array when possible, so it generally uses less additional memory.

For example, if a large image takes 500 MB of memory, using `flatten()` may require another 500 MB for the copied array. However, `ravel()` can usually work with the existing data without creating another 500 MB copy.

Therefore, when working with **large images and limited memory, `ravel()` is preferred because it can avoid creating an unnecessary copy and reduce additional memory usage.**

**Example:**
```python
image = np.array([[1, 2],
                  [3, 4]])

flat_image = image.flatten()
ravel_image = image.ravel()

print(flat_image)
print(ravel_image)

## C2. Standardize a 1D NumPy Array

**Question:**  
Write a function `standardize(x)` that takes a 1D NumPy array and returns its z-score standardized version using only NumPy. **[5 marks]**

````python
# Program: Standardize a 1D NumPy array using Z-score standardization
# Define a function to standardize the input array
def standardize(x):
    # Calculate the mean of the array
    mean = np.mean(x)
    # Calculate the standard deviation of the array
    std = np.std(x)
    # Apply the Z-score formula
    return (x - mean) / std
# Create a sample 1D NumPy array
scores = np.array([10, 20, 30, 40, 50])
# Standardize the scores
standardized_scores = standardize(scores)
# Display the result
print(standardized_scores)
````

### C2.Justification:

The `standardize()` function converts the values into Z-scores using the formula `(x - mean) / standard deviation`. First, `np.mean()` calculates the average value and `np.std()` calculates the standard deviation. Then, the formula is applied to every element of the array using NumPy vectorization, so no explicit loop is required.

## C4. Standardize a Dataset Column-Wise

**Question:**  
Given `X` with shape `(100, 5)`, compute the column-wise mean and standard deviation, and standardize `X` in one vectorized expression. **[5 marks]**

````python
# Program: Compute column-wise mean and standard deviation and standardize the dataset
# Create a sample dataset with 100 samples and 5 features
X = np.random.default_rng(42).normal(size=(100, 5))

# Calculate the mean and standard deviation for each column
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
# Standardize all columns using one vectorized expression
X_standardized = (X - mean) / std

# Display the shapes
print("Original shape:", X.shape)
print("Mean shape:", mean.shape)
print("Standardized shape:", X_standardized.shape)
````

### C4.Justification

Therefore, the standardized dataset keeps the original shape `(100, 5)` without using an explicit loop.

## C5. Linear Model Prediction and MSE
**Question:**  
Given input data `X`, weights `w`, bias `b`, and true values `y_true`, implement the prediction equation `y_pred = X @ w + b` and compute the Mean Squared Error (MSE) using `y_true`. **[5 marks]**

````python
# Program: Calculate linear model predictions and Mean Squared Error
# Create input data with 3 samples and 2 features
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

# Define the weight values for the two features
w = np.array([0.5, 1.0])
# Define the bias value
b = 2.0

# Calculate predictions using matrix multiplication
y_pred = X @ w + b

# Define the actual target values
y_true = np.array([4, 5, 6])

# Calculate Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)

# Display the predictions
print("Predictions:", y_pred)
# Display the Mean Squared Error
print("MSE:", mse)
````

### C5.Justification
`X @ w + b` calculates the predicted values using matrix multiplication and bias. MSE is the average of the squared differences between the actual and predicted values.

## C7. Normalize a Grayscale Image

**Question:**  
Given a grayscale image represented as a 2D NumPy array with pixel values in the range `[0, 255]`, apply min-max normalization to scale the values to the range `[0, 1]`. **[5 marks]**

````python
# Program: Normalize grayscale image pixel values to the range [0, 1]
# Create a sample grayscale image with pixel values from 0 to 255
image = np.array([[0, 50, 100],
                  [150, 200, 250],
                  [25, 125, 255]])

# Find the minimum pixel value
min_value = np.min(image)

# Find the maximum pixel value
max_value = np.max(image)

# Apply min-max normalization
normalized_image = (image - min_value) / (max_value - min_value)

# Display the original image
print("Original Image:")
print(image)

# Display the normalized image
print("\nNormalized Image:")
print(normalized_image)
# Display the minimum and maximum normalized values
print("\nMinimum:", np.min(normalized_image))
print("Maximum:", np.max(normalized_image))
````

### C7.Justification

Min-max normalization transforms the pixel values from their original range `[0, 255]` into the range `[0, 1]`. The formula `(image - min_value) / (max_value - min_value)` subtracts the minimum value and divides by the range of values. This is useful in machine learning and image processing because normalized values are easier for models to process.

## D1. Data Preprocessing Pipeline

**Question:**  
Given a raw 1D or 2D NumPy array containing `NaN` values, describe and implement a step-by-step preprocessing pipeline to:

1. Detect missing values.
2. Handle the missing values appropriately and justify the chosen strategy.
3. Normalize or standardize the cleaned data before using it in a model.

Implement the complete pipeline using a sample NumPy array containing at least one `NaN` value. **[8 marks]**

````python
# Program: Detect, handle, and standardize missing values in a NumPy array
# Create a sample dataset containing NaN values
data = np.array([10.0, 20.0, np.nan, 40.0, 50.0, np.nan])

# Step 1: Detect missing values
missing_values = np.isnan(data)

# Display the positions of missing values
print("Missing value positions:", missing_values)

# Count the total number of missing values
print("Number of missing values:", np.sum(missing_values))

# Step 2: Replace NaN values with the mean of the available values
mean_value = np.nanmean(data)
cleaned_data = np.where(
    np.isnan(data),
    mean_value,
    data
)
# Display the cleaned data
print("Cleaned data:", cleaned_data)

# Step 3: Standardize the cleaned data
mean = np.mean(cleaned_data)
std = np.std(cleaned_data)
standardized_data = (cleaned_data - mean) / std
# Display the standardized data
print("Standardized data:", standardized_data)
````

### D1.Justification

The preprocessing pipeline first uses `np.isnan()` to detect the missing values in the dataset. The missing values are replaced with the mean of the available values using `np.nanmean()`, because removing samples is unnecessary for this small dataset and mean imputation keeps all the data points. After handling the missing values, the cleaned data is standardized using the Z-score formula so that the values are centered around 0 with a standard deviation of 1. This preprocessing makes the data more suitable for many machine learning models.

## E1. Per-Column Min-Max Normalization

**Question:**  
Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns. **[5 marks]**

````python
# Program: Perform per-column Min-Max normalization using NumPy
# Create a 2D feature matrix
X = np.array([[10, 100, 1000],
              [20, 200, 2000],
              [30, 300, 3000],
              [40, 400, 4000]])

# Find the minimum value of each column
column_min = np.min(X, axis=0)

# Find the maximum value of each column
column_max = np.max(X, axis=0)

# Apply Min-Max normalization to each column
X_normalized = (X - column_min) / (column_max - column_min)

# Display the original matrix
print("Original Matrix:")
print(X)

# Display the normalized matrix
print("\nNormalized Matrix:")
print(X_normalized)
````

### Justification

`np.min(X, axis=0)` and `np.max(X, axis=0)` calculate the minimum and maximum values separately for each column. The Min-Max formula is then applied using NumPy broadcasting, which normalizes every feature independently to the range `[0, 1]`. Since the operation is vectorized, no explicit loops over rows or columns are required.

## E2. Three-Level Score Classification

**Question:**  
Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:

- **Fail** (< 40)
- **Average** (40–70 inclusive)
- **Excellent** (> 70)

**Hint:**  
You may need to nest or combine `np.where()` calls. **[5 marks]**

````python
# Program: Classify exam scores into Fail, Average, and Excellent
# Create a 1D array of exam scores
scores = np.array([25, 39, 40, 55, 70, 71, 85])

# Classify scores using nested np.where() without loops or if/else
classification = np.where(
    scores < 40,
    "Fail",
    np.where(scores <= 70, "Average", "Excellent")
)
# Display the scores and their classifications
print("Scores:", scores)
print("Classification:", classification)
````

### Justification

The nested `np.where()` checks each score against the given conditions. Scores below 40 are classified as `"Fail"`, scores from 40 to 70 are classified as `"Average"`, and scores above 70 are classified as `"Excellent"`. The entire classification is done using one vectorized expression without `if/else` statements or explicit loops.
