## Q1. NumPy Broadcasting
**Question:**

1. Two NumPy arrays have shapes `(3, 1)` and `(1, 3)`. Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?

**Ans:** The element-wise sum of arrays with shapes `(3, 1)` and `(1, 3)` will have the shape **`(3, 3)`**.

NumPy broadcasting allows arrays with compatible but different shapes to be used in arithmetic operations. Here:

- Array `a` has shape `(3, 1)`.
- Array `b` has shape `(1, 3)`.
- NumPy broadcasts both arrays to shape `(3, 3)` and performs element-wise addition.

Given:

```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

**1a. Predict the output of the given code.**

**Ans. predicted Output:**
```
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)
```
**1b. Identify the shapes of arrays `a` and `b` before addition.**

**Ans. Shapes of arrays `a` and `b` before addition:**
- Shape of `a` = (3, 1)
- Shape of `b` = (1, 3)

**1c.Explain how NumPy broadcasting enables the addition of these arrays.**

**Ans: Broadcasting:**

NumPy broadcasting allows arrays with compatible shapes to perform element-wise operations.

Here, `a` has shape (3, 1) and `b` has shape (1, 3). NumPy broadcasts them to a compatible shape of (3, 3) and performs the addition.


**1d. What would happen if `b` were replaced with `np.array([1, 2])`? Explain the reason.**

**Ans. If `b = np.array([1, 2])`:**

The operation will raise a `ValueError` because the shapes (3, 1) and (2,) are not compatible for broadcasting.

Therefore, the arrays cannot be added element-wise.

````python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b         #adds the two arrays together, broadcasting the single column of a across the three columns of b

print(result)
print(result.shape)    #displays the shape of the resulting array, which is (3, 3)
````

**D4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.**


````python
# Before converting Marks to numeric, identify the reason for missing marks
import numpy as np
df["missing_marks"] = np.where(
    df["Marks"] == "Absent",
    "Absent",
    np.where(
        df["Marks"].isna(),"not_entered","Marks"))
#cleaning
df["Marks"] = (
    df["Marks"]
    .astype("string")
    .str.strip()
    .replace({
        "Absent": pd.NA,
        "NA": pd.NA,
        "": pd.NA
    })
    .str.replace(" marks", "", regex=False)
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
display(df["Marks"])
````

**Explanation:** The Marks column is cleaned by removing extra spaces and text such as " marks", handling "Absent", "NA", and blank entries as missing values, and finally converting the column into a numeric format.
